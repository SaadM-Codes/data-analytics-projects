<ul>
<li>Before we can do anything with the salaries, we have to convert them into something more usable</li>
<ul>
<li>Note: the salaries can be yearly, monthly or hourly salaries</li>
<ul>
<li>We don't also need the Indian rupee –sign (₹)</li>
</ul>
<li>You can use the template in Moodle to help you out with this (Salary filtering, pandas exercise 3)</li>
</ul>
</ul>

```python
# removing the rupee sign as we don't need 

df['Salary'] = df['Salary'].str.replace('₹','')
df['Salary'] = df['Salary'].str.replace(',','')
df.head()
```

```python
def yearly_wage(row):
    # identifying if the salary is given hourly, monthly, or yearly.

    # spliting the salary data into two parts:
    split = row.split('/')

    # converting the salary amount to an integer (whole number).
    # this will allows us to do calculations with it.
    split[0] = int(split[0])

    # if the salary is hourly ('hr'):
    # multiplying the hourly wage by the average work hours per year in India.
    # (In India, the average annual work hours are approximately 2117.01 hours.)
    if split[1] == "hr":
        result = split[0] * 2117.01

    # if the salary is monthly ('mo'):
    # multiplying the monthly salary by 12 to get the yearly wage.
    elif split[1] == "mo":
        result = split[0] * 12

    # if the salary is already yearly ('yr'):
    else:
        result = split[0]

    # returning the calculated yearly wage.
    return result

# applying the yearly_wage function to the 'Salary' column.
# this will convert all salary values to their annual (yearly) equivalent.
df['Salary'] = df['Salary'].apply(yearly_wage)



# done with the help of my friend (jeremy)
```

<li>Finally, get the amount of rows grouped by each of the new Income Group –field values and print them out. <b>(value_counts() is handy here!)</b></li> 


```python
# loading the loans.csv file 

df = pd.read_csv('loans.csv')
```

```python
# created a column a new column called "Income Group"

df['Income Group'] = df['Loan ID'] + df['Customer ID']
```

```python
# function to categorize income levels
def income_group(income):
    
    if income <= 25000:
        return '$25k or less'  # low income group
    elif 25001 <= income <= 50000:
        return '$25k-$50k'  # lower middle-income group
    elif 50001 <= income <= 100000:
        return '$50k-$100k'  # middle-income group
    elif 100001 <= income <= 200000:
        return '$100k-$200k'  # upper middle-income group
    elif income >= 200001:
        return '$200k+'  # high-income group
    else:
        return 'empty'  # if income data is missing or invalid

# applying the income_group function to the 
# 'Annual Income' column and creating a new column 'Income Group'
df['Income Group'] = df['Annual Income'].apply(income_group)

# counting how many people fall into each income group and 
# displaying the results
print(df['Income Group'].value_counts())
```
