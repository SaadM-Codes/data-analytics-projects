```python

import xml.etree.ElementTree as et

# load XML-file -> get root node
xtree = et.parse("simple.xml")
node = xtree.getroot()

# column names for pandas when we create the final DataFrame
df_cols = ["id", "brand", "model", "year", "owned_by", "acquired", "price"]
rows = []

# parse "columns" or values in the XML
e_id = node.find("id").text
e_brand = node.find("brand").text
e_model = node.find("model").text
e_year = node.find("year").text
e_owned_by = node.find("owned_by").text
e_acquired = node.find("acquired").text
e_price = node.find("price").text


# create the DataFrame
rows.append({'id': e_id, 'brand': e_brand, 'model': e_model, 'year': e_year, 'owned_by': e_owned_by, 'acquired': e_acquired, 'price': e_price})
out_df = pd.DataFrame(rows, columns=df_cols)

# printing out_df
out_df
```

```python
# opening the file and reading
df = open("simple.json", "r")

data = df.read()

# using eval here because json is like an dictionary 
cars = eval(data)

# then i am calculating the sum of all cars prices
total_price = 0
for car in cars:
    total_price += car["price"]

# then after i will divide by the number of cars
average_price = total_price / len(cars)

# then the final step-> print the result
print("The average price is:", average_price)

# and on the end close the opened simple.json file
df.close()
```
